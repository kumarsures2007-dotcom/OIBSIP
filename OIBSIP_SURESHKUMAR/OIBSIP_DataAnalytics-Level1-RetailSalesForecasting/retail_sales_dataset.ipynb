Time series analysis

```python
# create month column
df["Month"]=df["Date"].dt.to_period("M")
df[["Date","Month"]].head()
```

```python
#Calculate Monthly Sales

monthly_sales=df.groupby("Month")["Total Amount"].sum()
monthly_sales
```

```python
#Plot Monthly sales trend
plt.figure(figsize=(12,5))
monthly_sales.plot(marker="o",linewidth=2)
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```

### Observation

* The line chart shows the monthly sales trend  over the available period

* Months with higher peaks indicate stronger sales perrformance.

* Months with lower sales may require promotional campaigns to improve revenue.

```python
# Quater Sales Trend
```

```python
#Create quater column
df["Quarter"]=df["Date"].dt.to_period("Q")
```

```python
quartely_sales=df.groupby("Quarter")["Total Amount"].sum()
quartely_sales

```

```python
#Plot Quartely Sales
plt.figure(figsize=(8,5))
quartely_sales.plot(marker="o",linewidth=2,color="green")
plt.title("Quartely Sales Trend")
plt.xlabel("Quarter")
plt.ylabel("Total Sales")
plt.show()
```

### observation

- The quartely sales chart summarizes revenue across each quarter.

- The quarter with the highest sales indicates the strongest business performance.

- Comaparing the quarters helps identify seasonal trends and plan inventory and marketing strategies.

# Customer demographics analysis: 

        * distribution of customer age groups
        * gender breakdown

```python
plt.figure(figsize=(8,5))
sns.histplot(df["Age"],bins=10,kde=True)
plt.title("Customer Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Customers")
plt.show()
```

### Observation

    # The histogram shows how customers are distributed across different age groups.

    # The age  group with highest frequency represents the primary customer segment.

    # This insight helps businesses target their marketing campaigns effectively.

```python
# Age Groups 
bins=[18,25,35,45,55,65]
labels=["18-25","26-35","36-45","46-55","56-65"]
df["Age Group"]=pd.cut(df["Age"],bins=bins,labels=labels)

df[["Age","Age Group"]].head()
```

```python
#Count customer in Each Age Group
age_group=df["Age Group"].value_counts().sort_index()
age_group
```

```python
#Plot age group distribution
plt.figure(figsize=(8,5))
sns.countplot(data=df,x="Age Group")
plt.title("Customer Age Group Distribution")
plt.xlabel("Age Group")
plt.ylabel("Number of Customers")
plt.show()
```

### Observation
- The chart shows the number of customers in each age group.

- The largest age group represents the most active customer segment.

- Businesses can design age-specific promotions based on this analysis.


```python
#Gender

gender_count = df["Gender"].value_counts()

gender_count
```

```python
#Plot Gender Distribution
plt.figure(figsize=(8,5))
sns.countplot(data=df,x="Gender")
plt.title("Group Distribution")
plt.xlabel("Gender",color="red")
plt.ylabel("Number of Customers",color="red")
plt.show()
```

### Observation

- The chart compares the number of male and female customers.

- A balanced distribution suggests similar customer participation from both genders.

- If one gender dominates, targeted marketing strategies can be developed.

# Product Analysis

```python
#Best-Selling Product Categories (By Quantity)

best_selling=df.groupby("Product Category")["Quantity"].sum().sort_values(ascending=False)

best_selling
```

```python
#Plot Best-Selling 
plt.figure(figsize=(8,5))
sns.barplot(
    x=best_selling.index,
    y=best_selling.values,
    hue=best_selling.index,
    palette="viridis",
    legend=False
    
)
plt.title("Best-Selling Product Categories")
plt.xlabel("Product Category")
plt.ylabel("Total Quantity Sole")
plt.show()
```

### Observation 

- The chart shows the total quantity sold for each product category.

- The category with the highest quantity sold is the most popular among customers

- Popular categories should be prioritized for inventory management.

```python
# Revenue by Product Category 
category_revenue=df.groupby("Product Category")["Total Amount"].sum().sort_values(ascending=False)
category_revenue
```

```python

plt.figure(figsize=(8,5))

sns.barplot(
    x=category_revenue.index,
    y=category_revenue.values,
    palette="magma"
)

plt.title("Revenue by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Revenue")

plt.show()
```

### Observation

- The bar chart compares the total revenue generated by each product category.
- The category with the highest revenue contributes the most to the business.
- Low-performing categories may require promotional offers or pricing strategies.

## Correlation Heatmap

```python
numeric_df=df.select_dtypes(include=["int64","float64"])
numeric_df.head()
```

```python
correlation=numeric_df.corr()
correlation
```

```python
#Plot Correlation Heatmap
plt.figure(figsize=(8,6))
sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)
plt.title("Correlation Heatmap")
```

### Observation

- The heatmap shows the correlation between numerical variables.
- Correlation values range from **-1 to +1**.
- A value close to **+1** indicates a strong positive relationship.
- A value close to **-1** indicates a strong negative relationship.
- A value close to **0** indicates little or no relationship.

```python
#Addition Visualization (Sales by gender)
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="Gender",
    y="Total Amount"
    ,palette="Set2",
     hue="Gender",
     legend=False
)

plt.title("Total Amount by Gender")
plt.xlabel("Gender")
plt.ylabel("Total Amount")

plt.show()
```

### Observation

- The box plot compares the distribution of purchase amounts for male and female customers.
- It shows the median, spread, and any outliers in spending.
- If one gender has a higher median spending, targeted marketing strategies can be designed for that customer segment.

```python
# Sales by Age Group
age_sales = df.groupby("Age Group")["Total Amount"].sum()

plt.figure(figsize=(8,5))

sns.barplot(
    x=age_sales.index,
    y=age_sales.values,
    palette="Blues_d",
    hue=age_sales.index,
    legend=False
)

plt.title("Total Sales by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Total Sales")

plt.show()
```

### Observation

- The chart shows which age group contributes the most revenue.
- This helps identify the most valuable customer segment.
- Marketing campaigns and product recommendations can be tailored for high-value age groups.
