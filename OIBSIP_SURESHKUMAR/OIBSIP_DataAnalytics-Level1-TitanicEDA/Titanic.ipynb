Detect Outliers Using the IQR Method


```python
# a)FInd Outliers in Age


Q1 = df["Age"].quantile(0.25)
Q3 = df["Age"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

age_outliers = df[(df["Age"] < lower) | (df["Age"] > upper)]

print("Number of Age Outliers:", len(age_outliers))
```

```python
#(b) Find Outliers in Fare
Q1 = df["Age"].quantile(0.25)
Q3 = df["Age"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

age_outliers = df[(df["Age"] < lower) | (df["Age"] > upper)]

print("Number of Age Outliers:", len(age_outliers))
```

# Handle Outliers (Capping)

```python
# Cap Age 
Q1 = df["Age"].quantile(0.25)
Q3 = df["Age"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df["Age"] = np.clip(df["Age"], lower, upper)
```

```python
# Cap Fare
Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df["Fare"] = np.clip(df["Fare"], lower, upper)
```

```python

plt.figure(figsize=(6,4))
sns.boxplot(x=df["Age"])
plt.title("Boxplot of Age")
plt.show()

plt.figure(figsize=(6,4))
sns.boxplot(x=df["Fare"])
plt.title("Boxplot of Fare")
plt.show()
```

### Observation 
Outlier Detection and Treatment
# The IQR (Interquartile Range) method was used to identify outliers.
# Outliers were mainly found in the Age and Fare columns.
# Instead of removing these records, outliers were capped using the IQR limits.
# Capping preserves all observations while reducing the impact of extreme values on analysis and machine learning models.
Observation
# Outliers were successfully identified using the IQR method.
# Extreme values were capped to reduce their influence.
# No rows were removed, so all observations remain available for analysis.
